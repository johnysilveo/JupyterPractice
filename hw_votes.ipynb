```python
votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]
counts = Counter(votes)
most_common = counts.most_common(1)
top_3 = counts.most_common(3)
rows = [
    [fruits, count]
    for fruits, count in counts.items()
]
headers = ["Fruits", "Count",]
counts_l = [counts]
ui.show_markdown_table(headers, rows)
ui.show_header("Top 3 most common fruits")
ui.show_result(counts.most_common(3))
ui.show_header("Most common fruits")
ui.show_result(counts.most_common(1))
```
