```python
import numpy as np
np.random.seed(42)
```

```python
WIDTH = 90
INNER_WIDTH = WIDTH - 2

def border() -> str:
    pattern = "<>"
    middle = (pattern * WIDTH)[:WIDTH - 2]
    return f"*{middle}*"

def centered(text: str = "") -> str:
    return f"|{text:^{INNER_WIDTH}}|"

def show_header(title: str) -> None:
  print(border())
  print(centered())
  print(centered(title))
  print(centered())
  print(border())

def result_border() -> str:
    return "+" + "-" * INNER_WIDTH + "+"

def show_result(value) -> None:
    print(result_border())
    for line in str(value).splitlines():
        print(centered(line))
    print(result_border())

a = np.arange(1, 6)
b = np.arange(6, 11)
show_header("a")
show_result(a)
show_header("b")
show_result(b)
concat = np.concatenate((a, b))
show_header("Конкатенація")
show_result(concat)

stack_rows = np.vstack((a, b))
show_header("5x2 (рядки)")
show_result(stack_rows)
stack_cols = np.column_stack((a, b))
show_header("5x2 (стовпці)")
show_result(stack_cols)
```
