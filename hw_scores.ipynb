```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]
math_avg = sum(math) / len(math)
math_min = min(math)
math_max = max(math)
physics_avg = sum(physics) / len(physics)
physics_min = min(physics)
physics_max = max(physics)
english_avg = sum(english) / len(english)
english_min = min(english)
english_max = max(english)
headers = ["Subject", "Average", "Min", "Max"]
rows = [
    ["Math", math_avg, math_min, math_max],
    ["Physics", physics_avg, physics_min, physics_max],
    ["English", english_avg, english_min, english_max]
]
ui.show_markdown_table(headers, rows)
averages = {
    "Math": math_avg,
    "Physics": physics_avg,
    "English": english_avg
}
best_subject = max(averages, key=averages.get)
best_avg = averages[best_subject]
headers1 = ["Best avg subject", "Score"]
rows1 = [
    [best_subject, best_avg]
]
print(ui.border())
ui.show_markdown_table(headers1, rows1)

```
