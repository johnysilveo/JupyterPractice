```python
import numpy as np
import cli_ui
```

```python
expenses = [1200,1500,800,600,1000,400,900]
total = sum(expenses)
avg_exp = sum(expenses) / len(expenses)
max_exp = max(expenses)
min_exp = min(expenses)
expenses = np.array(expenses)
more_1000 = expenses[expenses >= 1000]
rows = [
    [total, avg_exp, max_exp, min_exp, more_1000],
]
headers = ["Total", "Avg Exp", "Max Exp", "Min Exp", "More then 1000$ (Days)"]
cli_ui.show_markdown_table(headers, rows)
```
